```python
import pandas as pd
import matplotlib.pyplot as plt

data = {'Year': list(range(2015, 2026)),'App Downloads': [500, 750, 1100, 2200, 3000, 4200, 5800, 7500, 9200, 11500, 13000]}

df = pd.DataFrame(data)

plt.plot(df['Year'],df['App Downloads'],marker='o')
plt.title('Trend of Mobile Application Downloads')
plt.xlabel('Year')
plt.ylabel('Number of Downloads')
plt.show()
```
