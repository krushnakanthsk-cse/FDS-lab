```python
import pandas as pd

structured_data = pd.DataFrame({
    'Roll_No': [101, 102, 103],
    'Student': ['Arun', 'Kavin', 'Rahul'],
    'Marks': [85, 92, 78]
})

print("Structured Data:\n", structured_data)

unstructured_data = "Arun scored very well in Python and showed excellent programming skills."

print("\nUnstructured Data:\n", unstructured_data)

semi_structured_data = {
    'Roll_No': 101,
    'Student': 'Arun',
    'Details': {
        'Course': 'Computer Science',
        'Grade': 'A'
    }
}

print("\nSemi-structured Data:\n", semi_structured_data)
```
